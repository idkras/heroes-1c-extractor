## 1. Подключение к базе данных


```python
# Создаем экстрактор и blob процессор
extractor = FlatTableExtractor('../data/raw/1Cv8.1CD')
blob_processor = BlobProcessor()

# Подключаемся к базе
extractor.db_connector.connect()

print('✅ Подключение к базе данных установлено')
print(f'📊 База данных: {extractor.db_connector.file_path}')
```

## 2. Анализ _DOCUMENT138_VT3118 - Табличная часть поступлений


```python
print('📊 _DOCUMENT138_VT3118 - Табличная часть поступлений')
print('=' * 70)

try:
    # Получаем табличную часть
    table = extractor.db_connector.get_table('_DOCUMENT138_VT3118')
    print(f'📋 Размер таблицы: {len(table):,} записей')
    
    # Анализируем первые 10 записей
    count = 0
    for row in table:
        if count >= 10:
            break
            
        print(f'\n📝 Запись {count + 1}:')
        
        try:
            # Получаем данные строки как словарь
            row_dict = row.as_dict()
            
            # Импортируем функции маппинга
            from src.extract_all_available_data import get_field_display_name
            
            # Показываем ВСЕ поля записи с маппингом
            for field_name, value in row_dict.items():
                # Применяем маппинг полей
                display_name = get_field_display_name(field_name)
                
                if isinstance(value, bytes) and len(value) > 0:
                    print(f'  {display_name}: [BLOB {len(value)} байт]')
                elif value is not None:
                    # Ограничиваем длину для читаемости
                    str_value = str(value)
                    if len(str_value) > 100:
                        str_value = str_value[:100] + '...'
                    print(f'  {display_name}: {str_value}')
                else:
                    print(f'  {display_name}: None')
                        
        except Exception as e:
            print(f'  ❌ Ошибка обработки записи: {e}')
        
        count += 1
        
except Exception as e:
    print(f'❌ Ошибка анализа таблицы: {e}')
```

## 4. Создание плоской таблицы с маппингом полей


```python
# Создаем плоскую таблицу с маппингом полей
print('📊 СОЗДАНИЕ ПЛОСКОЙ ТАБЛИЦЫ С МАППИНГОМ ПОЛЕЙ')
print('=' * 70)

try:
    # Импортируем функции маппинга
    from src.extract_all_available_data import get_field_display_name
    
    # Создаем плоскую таблицу из первых 10 записей
    flat_table_data = []
    
    # Анализируем _DOCUMENT138_VT3118
    table = extractor.db_connector.get_table('_DOCUMENT138_VT3118')
    print(f'📋 Создаем плоскую таблицу из {min(10, len(table))} записей...')
    
    count = 0
    for row in table:
        if count >= 10:
            break
            
        try:
            # Получаем данные строки как словарь
            row_dict = row.as_dict()
            
            # Создаем запись с маппингом полей
            flat_record = {}
            for field_name, value in row_dict.items():
                # Применяем маппинг полей
                display_name = get_field_display_name(field_name)
                
                if isinstance(value, bytes) and len(value) > 0:
                    flat_record[display_name] = f'[BLOB {len(value)} байт]'
                elif value is not None:
                    flat_record[display_name] = str(value)
                else:
                    flat_record[display_name] = None
            
            flat_table_data.append(flat_record)
            count += 1
            
        except Exception as e:
            print(f'  ❌ Ошибка обработки записи {count + 1}: {e}')
            count += 1
    
    # Создаем DataFrame
    if flat_table_data:
        df = pd.DataFrame(flat_table_data)
        print(f'✅ Создана плоская таблица: {df.shape[0]} строк, {df.shape[1]} колонок')
        
        # Показываем DataFrame как таблицу (как на скриншоте)
        print('\n📊 ПЛОСКАЯ ТАБЛИЦА С МАППИНГОМ ПОЛЕЙ:')
        print('=' * 100)
        
        # Показываем первые 5 строк с первыми 8 колонками для читаемости
        display_df = df.head(5).iloc[:, :8]  # Первые 5 строк, первые 8 колонок
        
        # Форматируем вывод как таблицу
        print(display_df.to_string(index=True, max_cols=8, max_colwidth=20))
        
        print(f'\n📈 Полная таблица: {df.shape[0]} строк × {df.shape[1]} колонок')
        print(f'📋 Все колонки: {list(df.columns)}')
        
        # Показываем статистику по типам данных
        print(f'\n📊 СТАТИСТИКА ПО ТИПАМ ДАННЫХ:')
        print(df.dtypes.value_counts())
        
        # Показываем количество непустых значений
        print(f'\n📊 ЗАПОЛНЕННОСТЬ ПОЛЕЙ:')
        non_null_counts = df.count()
        for col in non_null_counts.head(10).index:
            print(f'  {col}: {non_null_counts[col]}/{len(df)} ({non_null_counts[col]/len(df)*100:.1f}%)')
    
except Exception as e:
    print(f'❌ Ошибка создания плоской таблицы: {e}')
    import traceback
    traceback.print_exc()
    if flat_table_data:
        df = pd.DataFrame(flat_table_data)
        print(f'✅ Создана плоская таблица: {df.shape[0]} строк, {df.shape[1]} колонок')
        print(f'📊 Колонки: {list(df.columns)[:10]}...')  # Показываем первые 10 колонок
        
        # Показываем первые 3 строки
        print('\n📋 ПЕРВЫЕ 3 СТРОКИ ПЛОСКОЙ ТАБЛИЦЫ:')
        print('=' * 70)
        for i, row in df.head(3).iterrows():
            print(f'\n📝 Строка {i + 1}:')
            for col, value in row.items():
                if pd.notna(value):
                    str_value = str(value)
                    if len(str_value) > 50:
                        str_value = str_value[:50] + '...'
                    print(f'  {col}: {str_value}')
    
except Exception as e:
    print(f'❌ Ошибка создания плоской таблицы: {e}')
```
